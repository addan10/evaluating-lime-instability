```python
import pandas as pd
from datasets import load_dataset
from sklearn.model_selection import train_test_split

dataset = load_dataset("FinanceMTEB/financial_phrasebank", split="train")
```

```python
texts = list(dataset["text"])
labels = list(dataset["label"])
class_names = ["Negative", "Neutral", "Positive"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=42, stratify=labels
)
```

```python
sample_idx = 10
sample_headline = X_test[sample_idx]
true_label = class_names[y_test[sample_idx]]

print(f'Target Headline: "{sample_headline}"')
print(f"Ground Truth: {true_label}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
```

```python
pipeline = make_pipeline(
    TfidfVectorizer(ngram_range=(1, 2), stop_words="english"),
    LogisticRegression(max_iter=1000, random_state=42),
)

pipeline.fit(X_train, y_train)
```

```python
pred_probs = pipeline.predict_proba([sample_headline])[0]
pred_class_idx = int(pipeline.predict([sample_headline])[0])
```

```python
print(f"\nModel Prediction: {class_names[pred_class_idx]}")
for idx, name in enumerate(class_names):
    print(f"  P({name}): {pred_probs[idx]:.4f}")
```

```python
from lime.lime_text import LimeTextExplainer

explainer = LimeTextExplainer(
    class_names=class_names, split_expression=r"\W+"  # Splits on spaces & punctuation
)
```

```python
def get_explanation(text, target_class, num_features=5, num_samples=250):
    exp = explainer.explain_instance(
        text_instance=text,
        classifier_fn=pipeline.predict_proba,
        labels=[target_class],
        num_features=num_features,
        num_samples=num_samples,
    )
    # Extract weights for the target class
    return dict(exp.as_list(label=target_class))
```

```python
runs_data = {}
all_features = set()

NUM_RUNS = 3
for run_id in range(1, NUM_RUNS + 1):
    weights = get_explanation(
        sample_headline,
        target_class=pred_class_idx,
        num_features=5,
        num_samples=250,
    )
    runs_data[f"Run {run_id}"] = weights
    all_features.update(weights.keys())

print(f"Total distinct tokens selected across all 3 runs: {len(all_features)}")
```

```python
import pandas as pd

table_rows = []
for token in sorted(all_features):
    row = {"Feature / Token": token}
    for run_id in range(1, NUM_RUNS + 1):
        run_name = f"Run {run_id}"
        if token in runs_data[run_name]:
            row[run_name] = f"{runs_data[run_name][token]:+.4f}"
        else:
            row[run_name] = "Not Selected"
    table_rows.append(row)

df_comparison = pd.DataFrame(table_rows)

print(f'\nHeadline: "{sample_headline}"')
print(f"Class Explained: {class_names[pred_class_idx]}")
print("\n--- LIME Feature Comparison Across 3 Runs ---")
print(df_comparison.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Convert text table into numbers (replace 'Not Selected' with 0)
clean_df = df_comparison.set_index("Feature / Token").replace(
    "Not Selected", 0
)
clean_df = clean_df.astype(float)

# Plot a grouped bar chart
fig, ax = plt.subplots(figsize=(8, 4))
clean_df.plot(kind="bar", ax=ax, width=0.7)

# Add title, labels, and zero line
plt.axhline(0, color="black", linewidth=0.8, linestyle="--")
plt.title("LIME Feature Weights Across 3 Runs")
plt.ylabel("Weight")
plt.xlabel("Words")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle=":", alpha=0.6)
plt.legend(title="Run")

plt.tight_layout()
plt.show()
```
